# Simple custom Tic-Tac-Toe AIs

```python
from tictactoe import TicTacToePlayer
import random
import numpy as np
```

```python
class RandomPlayer(TicTacToePlayer):
    def play(self, game):
        return random.choice(game.get_playable_cells())
```

```python
class DeterministicPlayer(TicTacToePlayer):
    def play(self, game):
        grid = game.get_state()
        # win if possible, else try not to lose
        for p in (1, -1):
            mask = grid == p
            for x, y in zip(*np.where(grid == game.CELL_EMPTY)):
                if np.sum(mask[x, :]) == 2 or np.sum(mask[:, y]) == 2:
                    return x, y
                if x == y and np.diag(mask).sum() == 2:
                    return x, y
                if x == 2 - y and np.diag(mask[:, ::-1]).sum() == 2:
                    return x, y
        # else play on the first available cell
        return game.get_playable_cells()[0]
```

# Precomputed exhaustive search strategy

```python
from tictactoe import TicTacToeRL
```

```python
def state_vector(game_state):
    return tuple(game_state.flatten())
```

```python
def find_Q_strategy():
    def explore(game):
        state = state_vector(game.get_state())
        # rewards for all possible actions in this state
        rewards = []
        for a in game.get_playable_cells():
            if (state, a) in Q:
                # if this combination has already been considered, use previously computed values
                reward = Q[state, a]
            else:
                # try this action and check the outcome
                copy = game.copy()
                copy.play(a)
                winner = copy.winner
                if winner is None:
                    # continue playing
                    # apply discounts for future moves
                    # alternate rewards between players
                    # (the provided reward is the opponent's)
                    reward = -alpha * explore(copy)
                elif winner == game.STATUS_FULL:
                    # draw
                    reward = 0
                else:
                    # the action just won the game
                    reward = 1
                Q[state, a] = reward
            rewards.append(reward)
        # we assume the player will try to maximize his reward
        return max(rewards)

    Q = {}
    alpha = 0.95  # can be set to 1 for tic-tac-toe (without erroneous moves), since the sum is finite
    explore(TicTacToeRL())
    return Q
```

```python
%%time
# Approx. 0.5 second exploration time
Q = find_Q_strategy()
print(f"Strategy has {len(Q)} states, with {len([v for v in Q.values() if v == 0])} ties, {len([v for v in Q.values() if v > 0])} wins and {len([v for v in Q.values() if v < 0])} losses")
```

```python
class PerfectPlayer(TicTacToePlayer):
    def __init__(self, Q):
        self.Q = Q

    def play(self, game):
        state = state_vector(game.get_state())
        actions = [(i, j) for i in range(3) for j in range(3)]
        best_action = []
        best_reward = -2
        for a in actions:
            if self.Q.get((state, a), -3) > best_reward:
                best_action = [a]
                best_reward = self.Q[state, a]
            elif self.Q.get((state, a), -3) == best_reward:
                best_action.append(a)
        if not best_action:
            raise ValueError("Invalid state: no valid action found in the strategy table")
        return random.choice(best_action)
```

```python
game = GameWidget(players=["human", PerfectPlayer(Q)])
game
```

```python
state = state_vector(game.get_state())
actions = [(i, j) for i in range(3) for j in range(3)]
for a in actions:
    if (state, a) in Q:
        print(a, Q[state, a])
```

# Random evaluation

```python
def evaluate(p1, p2, n=1000):
    scores = {"p1": 0, "p2": 0, "draw": 0}
    names = {1: "p1", 2: "p2", 3: "draw"}
    for _ in range(n):
        game = TicTacToeRL(player=random.choice([1, 2]))
        ended = False
        while not ended:
            if game.player == 1:
                ended = game.play(p1.play(game))
            else:
                ended = game.play(p2.play(game))
        scores[names[game.winner]] += 1
    return scores
```

```python
evaluate(RandomPlayer(), PerfectPlayer(Q))
```

```python
evaluate(PerfectPlayer(Q), PerfectPlayer(Q))
```

```python
evaluate(DeterministicPlayer(), PerfectPlayer(Q))
```

# Q-learning

**Principe** : entraîner un modèle d'apprentissage automatique pour fournir une approximation de la vraie fonction $Q(state, action) \to reward$.

Pour une transition découverte par exploration $(s, a) \to s'$ et donnant la récompense $r$, la table $Q$ est mise à jour par l'équation : $Q(s, a) \leftarrow r + \alpha \displaystyle\max_{a'} Q(s', a')$, c'est-à-dire que la récompense totale liée à l'action $a$ dans l'état $s$ est la somme de la récompense immédiate $r$ et du maximum des récompenses futures liées à l'état d'arrivée $s'$.
Les récompenses futures sont associées aux actions précédentes avec un coefficient de réduction $\alpha$ pour garantir la convergence.  
En pratique, cette rétro-propagation d'une récompense n'est pas faite immédiatement, car l'historique des actions n'est pas sauvegardée. La prise en compte des récompenses futures ne sera faite que lorsque l'on considérera à nouveau cet état par  exploration.

Pour un réseau de neurones, l'entrée du modèle est généralement un état $s$. Les neurones de la couche de sortie correspondent à chacune des actions possibles, avec comme valeur associée une estimation de la récompense (totale) liée à l'action $\hat{r}$.  
La différence entre la valeur $\hat{r}$ fournie par le modèle et la valeur $r + \alpha \displaystyle\max_{a'} Q(s', a')$ calculée sert à déterminer une erreur pour le mettre à jour.

**Considérations pratiques**

- Lors de l'apprentissage, il existe un compromis entre suivre des actions dont la récompense est favorable (on parle d'exploitation), et suivre une action moins intéressante, mais potentiellement liée à une récompense plus grande non encore découverte (on parle d'exploration). La manière usuelle de régler ce compromis est d'avoir une probabilité $\epsilon$ de réaliser une action aléatoire, et sinon de suivre la stratégie optimale d'après la table Q actuelle.
- Il est souvent (relativement) coûteux de consulter l'environnement pour calculer la transition $(s, a) \to (r, s')$. Il est donc souhaitable de la sauvegarder dans une base d'expériences précédentes (*experience replay buffer*) qu'on pourra réutiliser régulièrement (avec un échantillonnage aléatoire) pour l'entrainement du modèle. Notez que ces transitions ne dépendent pas de $Q$ et, dans un environnement déterministe, ne changent jamais.
- Lorsque l'IA doit jouer contre un adversaire, il est possible de conserver une ou plusieurs versions précédentes du modèle pour jouer ce rôle. Malheureusement, atteindre un modèle à peu près correct peut prendre un certain temps si tous les adversaires jouent mal, ce qui est souvent le cas au début de l'entraînement.
