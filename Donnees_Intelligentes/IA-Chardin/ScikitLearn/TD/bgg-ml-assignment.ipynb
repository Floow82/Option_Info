# Rappels

**Question** : Découper le jeu de données en deux sous-ensembles, X et y.

On pourra distinguer les informations pouvant être renseignées dès la sortie d'un jeu (et qui pourront servir d'entrées du modèle), des informations qui font intervenir les avis des utilisateurs.

```python
user_cols = [
    "usersrated",
    "average",
    "bayesaverage",
    "stddev",
    "owned",
    "trading",
    "wanting",
    "wishing",
    "numcomments",
    "numweights",
    "averageweight",
]
info_cols = list(games.columns.drop(user_cols))
info_cols
```

```python
#La sortie voulue
y = games["bayesaverage"]

#Tout le reste pour l'entrée
##X = games.drop(columns=["bayesaverage"])
X = games[info_cols]
```

**Question** : Dans le cadre d'un processus de validation basé sur un découpage préalable entrainement-test, découper le jeu de données en deux quatre variables, X_train, X_test, y_train et y_test à l'aide de la fonction [train_test_split](https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.train_test_split.html) de scikit-learn.  
À quoi ces quatre ensembles vont-ils servir ?

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20)
```

**Question** : Entraîner un modèle de type [*Support Vector Regressor (SVR)*](https://scikit-learn.org/stable/modules/generated/sklearn.svm.SVR.html).  
Remarque : ce type de modèle ne supporte pas les valeurs manquantes, et est sensible à la normalisation des données d'entrée.  
Votre solution devra prendre la forme d'un pipeline.

## **voir page 5 du cours** ##

```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.svm import SVR

# Spécifier les colonnes numériques
numeric_features = X.select_dtypes(include=['int64', 'float64']).columns

# Créer le préprocesseur pour les données numériques
numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='mean')),  # Remplir les valeurs manquantes avec la moyenne
    ('scaler', StandardScaler())  # Normaliser les données
])

# Créer le préprocesseur global
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_features)  # Appliquer le préprocesseur aux colonnes numériques
    ])

# Créer le pipeline
pipeline = Pipeline(steps=[('preprocessor', preprocessor),
                           ('regressor', SVR())])
```

**Question** : Visualiser les prédictions du modèle par rapport aux vraies notes sur le jeu de test.  
Identifier et calculer une métrique appropriée (un score).

```python
# Entraîner le modèle
pipeline.fit(X_train, y_train)

# Évaluer le modèle
train_score = pipeline.score(X_train, y_train)
test_score = pipeline.score(X_test, y_test)

print("Score d'entraînement:", train_score)
print("Score de test:", test_score)
```

**Question** : comparer ce résultat à celui obtenu avec un [*DummyRegressor*](https://scikit-learn.org/stable/modules/generated/sklearn.dummy.DummyRegressor.html) de scikit-learn.

```python
from sklearn.dummy import DummyRegressor

# Créer et entraîner le Dummy Regressor
dummy_regressor = DummyRegressor(strategy="mean")  # DummyRegressor utilisant la stratégie de prédiction basée sur la moyenne
dummy_regressor.fit(X_train, y_train)

# Évaluer les performances du Dummy Regressor
dummy_train_score = dummy_regressor.score(X_train, y_train)
dummy_test_score = dummy_regressor.score(X_test, y_test)

print("Score d'entraînement du Dummy Regressor:", dummy_train_score)
print("Score de test du Dummy Regressor:", dummy_test_score)

# Évaluer les performances du modèle SVR
train_score = pipeline.score(X_train, y_train)
test_score = pipeline.score(X_test, y_test)

print("Score d'entraînement du modèle SVR:", train_score)
print("Score de test du modèle SVR:", test_score)
```

**Question** : Peut-on améliorer la performance du modèle en intégrant les avis des utilisateurs (en excluant bien entendu la colonne à prédire) ?

Oui, l'intégration des avis des utilisateurs, en utilisant les autres colonnes comme caractéristiques pour votre modèle, pourrait potentiellement améliorer sa performance. Cependant, il est important de choisir judicieusement les caractéristiques à inclure dans le modèle, car l'inclusion de caractéristiques non pertinentes ou redondantes peut entraîner un surajustement et réduire les performances du modèle.

# Inspection du modèle

Certains modèles sont considérés comme des *boîtes noires*, c'est-à-dire qu'il est difficile pour un humain d'appréhender son fonctionnement et son processus de détermination de la sortie.

Une partie de la recherche en machine learning s'intéresse à l'explicabilité des modèles, avec quelques techniques qui peuvent s'appliquer aussi aux boîtes noires.

Une de ces techniques consiste à observer l'effet d'un [mélange des valeurs d'une colonne](https://christophm.github.io/interpretable-ml-book/feature-importance.html) sur l'erreur obtenue en sortie.

```python
from sklearn.inspection import permutation_importance
```

```python
importances = permutation_importance(pipeline, X_test, y_test, n_repeats=3, n_jobs=-1)
```
