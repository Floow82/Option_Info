# Génération automatique de *tags*

Pour cet exercice, on cherche à proposer automatiquement des tags à associer à un jeu à partir de sa description.

```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
```

```python
games = pd.read_csv("bgg/games.csv")
links = pd.read_csv("bgg/links.csv")
```

```python
links[["type", "value"]].value_counts().head(5)
```

Une première version s'intéresse uniquement au tag le plus commun : *card game*.

```python
games["card game"] = games["id"].isin(links[links["value"] == "Card Game"]["id"])
```

```python
X = games["description"]  # vecteur 1D
y = games["card game"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y)
```

## Modèle de référence (scikit-learn)

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics import ConfusionMatrixDisplay
from sklearn.pipeline import Pipeline
from sklearn.svm import SVC
```

```python
pipeline = Pipeline([
    ("tokenization", CountVectorizer()),
    ("classification", SVC(class_weight="balanced")),
])
```

```python
pipeline.fit(X_train, y_train)
```

```python
y_pred = pipeline.predict(X_test)
```

```python
ConfusionMatrixDisplay.from_predictions(y_test, y_pred)
plt.show()
```

# Sentence Transformers

Plusieurs dépôts de transformateurs proposent des [modèles pré-entraînés](https://www.sbert.net/docs/pretrained_models.html).

```python
from sentence_transformers import SentenceTransformer

mpnet = SentenceTransformer("all-mpnet-base-v2")
miniml = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
examples = [
    "Machine learning sometimes is quite hard to understand.",
    "I know nothing about artificial intelligence.",
    "Sometimes, this is a hard machine, you understand?",
    "The sky is blue.",
]
```

```python
from itertools import combinations
from sklearn.metrics.pairwise import cosine_similarity
```

```python
matching = []
for s1, s2 in combinations(examples, 2):
    e1 = mpnet.encode(s1)
    e2 = mpnet.encode(s2)
    sim = cosine_similarity(e1.reshape(1, -1), e2.reshape(1, -1))[0, 0]
    matching.append((sim, s1, s2))

for sim, s1, s2 in sorted(matching, reverse=True):
    print(f"{sim:.3f}: '{s1}' vs '{s2}'")
```

```python
mpnet.encode(["Machine learning sometimes is quite hard to understand."])
```

```python
%%time
# temps d'inférence sur GPU (RTX A4000)
# 1min 40s pour all-mpnet-base-v2
# 17s pour all-MiniLM-L6-v2
X_train_st = miniml.encode(X_train.to_list())
X_test_st = miniml.encode(X_test.to_list())
```

```python
clf = SVC(class_weight="balanced")
```

```python
clf.fit(X_train_st, y_train)
```

```python
y_pred = clf.predict(X_test_st)
```

```python
ConfusionMatrixDisplay.from_predictions(y_test, y_pred)
plt.show()
```

# BERT

```python
from transformers import BertModel, BertTokenizer

tokenizer = BertTokenizer.from_pretrained('bert-base-uncased', do_lower_case=True)
bert = BertModel.from_pretrained("bert-base-uncased", output_attentions=False, output_hidden_states=False)
bert.eval();  # disable dropout
```

### Exemples de transformations

**Tokenisation**

[CLS] est un token spécial ajouté en début de chaque texte. Comme BERT génère un vecteur par token, celui-ci sert principalement à avoir un token vide de sens permettant au modèle de générer un vecteur portant sur la phrase entière.

[SEP] est un token spécial pouvant séparer différentes parties d'un texte. Celui-ci est utilisé en particulier pour certaines tâches d'entraînement du modèle : prédire si deux parties peuvent être consécutives ou non.

```python
tokenizer(["Machine learning sometimes is quite hard to understand."], truncation=True, padding=True, return_token_type_ids=False, return_tensors="pt")
```

```python
for t in [ 101, 3698, 4083, 2823, 2003, 3243, 2524, 2000, 3305, 1012,  102]:
    print(tokenizer.convert_ids_to_tokens(t))
```

**Embedding**

Le modèle génère, en récupérant les valeurs de sa dernière couche cachée, un vecteur de taille 768 par token.  
BERT supporte jusqu'à 512 tokens en entrée (soit significativement moins que les LLMs récents, qui en ont plusieurs milliers), ce qui fait qu'en sortie on obtient une matrice de taille maximum 512 $\times$ 768.

Il existe plusieurs stratégies pour obtenir un vecteur par texte :
- considérer uniquement celui associé au premier token ([CLS]),
- calculer la moyenne des vecteurs,
- calculer le max des vecteurs.

```python
text = "Machine learning sometimes is quite hard to understand."
tokens = tokenizer(text, truncation=True, padding=True, return_token_type_ids=False, return_tensors="pt")
embeddings = bert(**tokens)
```

```python
e = embeddings.last_hidden_state.detach()
print(e.shape)
e
```

```python
# vecteur associé au premier token
e[0, 0]
```

```python
# moyenne des vecteurs
e[0].mean(axis=0)
```

```python
def embed(text, model):
    tokens = tokenizer(text, truncation=True, padding=True, return_token_type_ids=False, return_tensors="pt")
    embeddings = model(**tokens)
    return embeddings.last_hidden_state.detach()[0, 0]  # premier
    # return embeddings.last_hidden_state.detach()[0].mean(axis=0)  # moyenne
```

```python
matching = []
for s1, s2 in combinations(examples, 2):
    e1 = embed(s1, bert)
    e2 = embed(s2, bert)
    sim = cosine_similarity(e1.reshape(1, -1), e2.reshape(1, -1))[0, 0]
    matching.append((sim, s1, s2))

for sim, s1, s2 in sorted(matching, reverse=True):
    print(f"{sim:.3f}: '{s1}' vs '{s2}'")
```

```python
%%time
# temps d'inférence sur GPU : 1min 42s
X_train_bert = X_train.apply(embed, model=bert)
```

```python
%%time
# temps d'inférence sur GPU : 32s
X_test_bert = X_test.apply(embed, model=bert)
```

```python
clf = SVC(class_weight="balanced")
```

```python
clf.fit(X_train_bert.apply(pd.Series), y_train)
```

```python
y_pred = clf.predict(X_test_bert.apply(pd.Series))
```

```python
ConfusionMatrixDisplay.from_predictions(y_test, y_pred)
plt.show()
```

# Fine tuning de BERT avec PyTorch

```python
import torch
import torch.nn as nn
from torch.optim import AdamW
from torch.utils.data import Dataset, DataLoader, random_split
from transformers import BertForSequenceClassification, BertTokenizer
```

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(device)
```

### Transformation des entrées

```python
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased', do_lower_case=True)
```

```python
class BGGDataset(Dataset):
    def __init__(self, encodings, labels):
        self.encodings = encodings
        self.labels = labels.to_numpy().astype(int)

    def __getitem__(self, idx):
        item = {key: val[idx].clone().detach().to(device) for key, val in self.encodings.items()}
        item['labels'] = torch.tensor(self.labels[idx]).to(device)
        return item

    def __len__(self):
        return len(self.labels)
```

```python
X_train_bert = tokenizer(X_train.to_list(), truncation=True, padding=True, return_token_type_ids=False, return_tensors="pt")
```

```python
train_dataset = BGGDataset(X_train_bert, y_train)
```

```python
train_size = int(0.9 * len(X_train))
```

```python
train_subset, val_subset = random_split(train_dataset, [train_size, len(X_train) - train_size])
```

```python
model = BertForSequenceClassification.from_pretrained(
    "bert-base-uncased",
    num_labels=2, # nombre de labels en sortie (classification multi-classe possible)
    output_attentions=False,
    output_hidden_states=False,
)
model.to(device)
```

```python
criterion = nn.CrossEntropyLoss()  
optimizer = AdamW(model.parameters(), lr=2e-5)
```

```python
from tqdm import tqdm
```

```python
def train(model, optimizer, train_subset, criterion):
    train_loader = DataLoader(train_subset, batch_size=16, shuffle=True)
    model.train()
    total_loss = 0

    for batch in tqdm(train_loader):
        optimizer.zero_grad()
        input_ids = batch["input_ids"]
        attention_mask = batch["attention_mask"]
        labels = batch["labels"]
        outputs = model(input_ids, attention_mask=attention_mask, labels=labels)
        loss = outputs[0]
        loss.backward()
        optimizer.step()
        total_loss += loss.item()
    print(f'Training loss: {total_loss / len(train_loader)}')
```

```python
def evaluate(model, val_subset, criterion):
    val_loader = DataLoader(val_subset, batch_size=16)
    model.eval()
    total_loss = 0
    total_acc = 0

    with torch.no_grad():
        for batch in tqdm(val_loader):
            input_ids = batch["input_ids"]
            attention_mask = batch["attention_mask"]
            labels = batch["labels"]
            outputs = model(input_ids, attention_mask=attention_mask, labels=labels)
            loss = outputs[0]
            total_loss += loss.item()
            predictions = torch.argmax(outputs.logits, dim=1)
            total_acc += (predictions == labels).sum().item()

    print(f"Validation loss: {total_loss / len(val_loader)} Validation accuracy: {total_acc / len(val_subset):%}")
```

```python
# temps d'entraînement sur GPU : 25min
for epoch in range(4):
    train(model, optimizer, train_subset, criterion)
    evaluate(model, val_subset, criterion)
```

```python
# results for four epochs
training_loss = [0.3878422547131777, 0.27863580055324544, 0.18750024814113614, 0.10779620860675125]
validation_loss = [0.2674279559536704, 0.2491668083571962, 0.2699549046145486, 0.36963754873639637]
validation_accuracy = [0.88435374, 0.89342404, 0.89909297, 0.89115646]
```

```python
model = BertForSequenceClassification.from_pretrained(
    "bert-base-uncased",
    num_labels=2,
    output_attentions=False,
    output_hidden_states=False,
)
model.load_state_dict(torch.load("bert-model/card-epoch-3.pt"))
model.to(device)
```

```python
X_test_bert = tokenizer(X_test.to_list(), truncation=True, padding=True, return_token_type_ids=False, return_tensors="pt").to(device)
test_dataset = BGGDataset(X_test_bert, y_test)
```

```python
y_pred = []
for batch in tqdm(DataLoader(test_dataset, batch_size=8)):
    input_ids = batch["input_ids"]
    attention_mask = batch["attention_mask"]
    labels = batch["labels"]
    prediction = model(input_ids, attention_mask=attention_mask, labels=labels)
    y_pred += prediction.logits.argmax(axis=1).tolist()
```

```python
ConfusionMatrixDisplay.from_predictions(y_test, y_pred)
plt.show()
```
